# This notebook demonstrates the way to use the STAC catalog that is created by Data Discovery Tool, for the data produced by the downstream applications of the destination Earth Twin engine, at different HPC centers and using differnt ESMs.

```python
from pystac_client import Client

# set pystac_client logger to DEBUG to see API calls
import logging
logging.basicConfig()
logger = logging.getLogger('pystac_client')
logger.setLevel(logging.DEBUG)
```

# Open the master STAC catalog and create a catalog object which can be queried to load the contents needed by user analytical tasks.

```python
# STAC API root URL
#URL = 'https://planetarycomputer.microsoft.com/api/stac/v1'
URL='/work/bm0146/k204247/DestinE/destinE_DDTool/DDTMasterSTACCatalog/catalog.json'

# custom headers
headers = []

cat = Client.open(URL, headers=headers)
cat
```

```python
for collection in cat.get_children():
    print(f"{collection}\n")
    for child in collection.get_children():
        print(f"  {child}")
        for grand_child in child.get_children():
            print(f"   {grand_child}")
            for greatgrand_child in grand_child.get_children():
                print(f"    {greatgrand_child}")
                for greatgreatgrand_child in grand_child.get_children():
                    print(f"     {greatgreatgrand_child}")
    
```
